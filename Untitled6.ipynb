```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline
import numpy as np

df = pd.read_excel('/content/dataset_lemmatized.xlsx')

df_clean = df.dropna(subset=["TEXT", "labels"])
X = df_clean["TEXT"]
y = df_clean["labels"]

# TF-IDF + Logistic Regression
tfidf_pipe = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=5000)),
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])

# 5-Fold Cross-validation
tfidf_scores = cross_val_score(tfidf_pipe, X, y, cv=5, scoring="f1")
print("TF-IDF F1 Score Mean:", np.mean(tfidf_scores))
print("TF-IDF F1 Score Std Dev:", np.std(tfidf_scores))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# BoW + Logistic Regression
bow_pipe = Pipeline([
    ("bow", CountVectorizer(binary=True, max_features=5000)),  # binary BoW
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])

# 5-Fold Cross-validation
bow_scores = cross_val_score(bow_pipe, X, y, cv=5, scoring="f1")
print("BoW F1 Score Mean:", np.mean(bow_scores))
print("BoW F1 Score Std Dev:", np.std(bow_scores))
```

```python
from gensim.models import Word2Vec
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.pipeline import make_pipeline
import numpy as np

# Tokenizer
tokenized_sentences = [text.split() for text in X]

# Word2Vec
w2v_model = Word2Vec(sentences=tokenized_sentences, vector_size=100, window=5, min_count=2, workers=4, seed=42)

# Embedding texts
def get_average_vector(tokens, model):
    valid_tokens = [token for token in tokens if token in model.wv]
    if not valid_tokens:
        return np.zeros(model.vector_size)
    return np.mean(model.wv[valid_tokens], axis=0)

X_w2v = np.array([get_average_vector(tokens, w2v_model) for tokens in tokenized_sentences])

# 4. Logistic Regression + 5-fold CV
clf = LogisticRegression(max_iter=1000, random_state=42)
scores = cross_val_score(clf, X_w2v, y, cv=5, scoring='f1')

print("Word2Vec F1 Score Mean:", np.mean(scores))
print("Word2Vec F1 Score Std Dev:", np.std(scores))
```

```python
from gensim.models import KeyedVectors
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
import numpy as np
from tqdm import tqdm

# Law2Vec upload
law2vec = KeyedVectors.load_word2vec_format('/content/Law2Vec.200d.txt', binary=False)

# Tokenizer
tokenized_sentences = [text.split() for text in X]

# Vectors
def get_law2vec_avg(tokens, model):
    valid = [model[token] for token in tokens if token in model]
    if not valid:
        return np.zeros(model.vector_size)
    return np.mean(valid, axis=0)

X_law2vec = np.array([get_law2vec_avg(tokens, law2vec) for tokens in tqdm(tokenized_sentences)])

# Model + CV
clf = LogisticRegression(max_iter=1000, random_state=42)
scores = cross_val_score(clf, X_law2vec, y, cv=5, scoring='f1')

print("Law2Vec F1 Score Mean:", np.mean(scores))
print("Law2Vec F1 Score Std Dev:", np.std(scores))

```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.model_selection import cross_val_score
from lightgbm import LGBMClassifier
from scipy.sparse import hstack

# 1. BoW (binary) temsili
bow_vectorizer = CountVectorizer(binary=True, max_features=5000)
X_bow = bow_vectorizer.fit_transform(X)

# 2. Yalnızca "güvenli" bonus özellikler
bonus_df = pd.DataFrame()
bonus_df["importance"] = df_clean["IMPORTANCE"]
bonus_df["respondent"] = df_clean["RESPONDENT"]
bonus_df["branch"] = df_clean["BRANCH"]

# 3. Preprocessing (One-hot ve Scaling)
cat_cols = ["respondent", "branch"]
num_cols = ["importance"]

preprocessor = make_column_transformer(
    (OneHotEncoder(handle_unknown='ignore'), cat_cols),
    (StandardScaler(), num_cols)
)

X_bonus = preprocessor.fit_transform(bonus_df)

# 4. BoW + Ekstra Özellikleri Birleştir
X_final = hstack([X_bow, X_bonus])

# 5. LightGBM modeli
model = LGBMClassifier(n_estimators=100, random_state=42, n_jobs=-1)

# 6. 5-Fold CV
scores = cross_val_score(model, X_final, y, cv=5, scoring='f1')

print("LightGBM (safe bonus) F1 Score Mean:", np.mean(scores))
print("LightGBM (safe bonus) F1 Score Std Dev:", np.std(scores))
```

```python
from transformers import BertTokenizer, BertForSequenceClassification
from transformers import Trainer, TrainingArguments
from datasets import Dataset
import torch
from sklearn.metrics import f1_score
from tqdm import tqdm

# Cihaz ayarı
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Tokenizer ve model yükleme
model_id = "nlpaueb/bert-base-uncased-echr"
tokenizer = BertTokenizer.from_pretrained(model_id)
model = BertForSequenceClassification.from_pretrained(model_id, num_labels=2).to(device)

# Veri hazırlığı
def preprocess_function(examples):
    return tokenizer(examples["TEXT"], padding="max_length", truncation=True, max_length=512)

df = pd.read_excel("/content/dataset_lemmatized.xlsx")  # kendi yolun
df = df[["TEXT", "labels"]].dropna()
dataset = Dataset.from_pandas(df)
dataset = dataset.map(preprocess_function, batched=True)

# Inference
model.eval()
all_preds, all_labels = [], []

for batch in tqdm(dataset.iter(batch_size=16)):
    input_ids = torch.tensor(batch['input_ids']).to(device)
    attention_mask = torch.tensor(batch['attention_mask']).to(device)
    labels = torch.tensor(batch['labels']).to(device)

    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        preds = torch.argmax(outputs.logits, dim=-1)

    all_preds.extend(preds.cpu().numpy())
    all_labels.extend(labels.cpu().numpy())

# Skor hesaplama
macro_f1 = f1_score(all_labels, all_preds, average='macro')
print("📢 ECHR-BERT Macro F1 Score:", macro_f1)
```

```python
# 🚀 HuggingFace Transformer Karşılaştırma - ECHR Dataset
import pandas as pd
import numpy as np
import torch
import evaluate
import time
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
)
from datasets import Dataset

# 📥 Excel'den veri oku (aynı klasöre yükle: dataset_lemmatized.xlsx)
df = pd.read_excel("dataset_lemmatized.xlsx")
df = df.dropna(subset=["TEXT", "labels"])
texts = df["TEXT"].astype(str).tolist()
labels = df["labels"].tolist()

# 🔀 Train/test ayır
train_texts, test_texts, train_labels, test_labels = train_test_split(
    texts, labels, test_size=0.2, stratify=labels, random_state=42
)

# 🤖 Dataset'e çevir
train_ds = Dataset.from_dict({"text": train_texts, "label": train_labels})
test_ds = Dataset.from_dict({"text": test_texts, "label": test_labels})

# 🔢 Kullanılacak modeller
models = {
    "BERT": "bert-base-uncased",
    "Legal-BERT-Base": "nlpaueb/legal-bert-base-uncased",
    "Legal-BERT-Small": "nlpaueb/legal-bert-small-uncased",
    "ECHR-BERT": "nlpaueb/bert-base-uncased-echr"
}

# 🔍 Preprocessing
def preprocess(tokenizer):
    def tokenize(batch):
        return tokenizer(batch["text"], padding="max_length", truncation=True, max_length=512)
    return tokenize

# 📊 Metrikler
accuracy_metric = evaluate.load("accuracy")

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    preds = np.argmax(predictions, axis=1)
    acc = accuracy_metric.compute(predictions=preds, references=labels)
    report = classification_report(labels, preds, output_dict=True)
    return {
        "accuracy": acc["accuracy"],
        "precision": report["weighted avg"]["precision"],
        "recall": report["weighted avg"]["recall"],
        "f1": report["weighted avg"]["f1-score"],
    }

# 📉 Confusion Matrix plot
def plot_confusion(cm, model_name):
    plt.figure(figsize=(4, 3))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=["Non-Violate", "Violate"],
                yticklabels=["Non-Violate", "Violate"])
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"Confusion Matrix - {model_name}")
    plt.tight_layout()
    plt.savefig(f"{model_name}_conf_matrix.png")
    plt.close()

# 🔁 Eğitim + değerlendirme
results = {}

for name, model_name in models.items():
    print(f"\n🔥 Training {name}...")

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    tokenized_train = train_ds.map(preprocess(tokenizer), batched=True)
    tokenized_test = test_ds.map(preprocess(tokenizer), batched=True)

    model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)

    training_args = TrainingArguments(
        output_dir=f"./results/{name}",
        per_device_train_batch_size=8,
        per_device_eval_batch_size=8,
        num_train_epochs=3,
        logging_steps=10,
        report_to="none"
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_train,
        eval_dataset=tokenized_test,
        compute_metrics=compute_metrics,
        tokenizer=tokenizer
    )

    trainer.train()
    eval_results = trainer.evaluate()

    # ⏱️ Inference Time
    inputs = tokenizer(test_texts[0], return_tensors="pt", truncation=True, padding=True).to(model.device)
    start = time.time()
    with torch.no_grad():
        _ = model(**inputs)
    single_time = time.time() - start

    # 🔍 Confusion matrix
    preds = trainer.predict(tokenized_test).predictions
    preds_cls = np.argmax(preds, axis=1)
    cm = confusion_matrix(test_labels, preds_cls)
    plot_confusion(cm, name)

    results[name] = {
        "metrics": eval_results,
        "inference_time_single": single_time,
        "confusion_matrix": cm
    }

# 🧾 Performans Tablosu
for model, data in results.items():
    m = data["metrics"]
    print(f"\n📊 {model} → Acc: {m['eval_accuracy']:.3f}, Prec: {m['eval_precision']:.3f}, Rec: {m['eval_recall']:.3f}, F1: {m['eval_f1']:.3f}, Inference: {data['inference_time_single']*1000:.1f} ms")
    print(f"Confusion Matrix:\n{data['confusion_matrix']}")
```
